```python
import json
import numpy as np
from pathlib import Path

root = Path(
    "/home/jovyan/thesis_project/FM_thesis/nnUNet/"
    "test_predictions/ID_nnUNet_preds/"
)

fold_means = []

for fold in range(5):
    path = root / f"fold_{fold}" / "summary.json"

    with open(path) as f:
        results = json.load(f)

    dice_values = []

    for case in results["metric_per_case"]:
        metrics = case["metrics"]["1"]

        # Evaluate only cases with an annotated lesion.
        if metrics["n_ref"] > 0:
            dice_values.append(metrics["Dice"])

    mean_dice = np.mean(dice_values)
    fold_means.append(mean_dice)

    print(
        f"Fold {fold}: "
        f"Dice = {mean_dice:.4f}, "
        f"Cases = {len(dice_values)}"
    )

print("\nOverall across folds:")
print(f"Mean Dice: {np.mean(fold_means):.4f}")
print(f"Std Dice:  {np.std(fold_means, ddof=1):.4f}")
```
