```python
malignant_patient_ids = (
    df.groupby("PatientID")
      .filter(lambda g: (
          (g["AbnormalityType"] == "Mass") &
          (g["PatientGroundtruth"] == "Malignant")
      ).any())
      ["PatientID"]
      .unique()
)


benign_patient_ids = (
    df.groupby("PatientID")
      .filter(lambda g: (
          (g["AbnormalityType"] == "Mass") &
          (g["PatientGroundtruth"] == "Benign")
      ).any())
      ["PatientID"]
      .unique()
)




len(benign_patient_ids)
```
