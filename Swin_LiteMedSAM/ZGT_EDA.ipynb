### How many patients have malignant tumors and how many benign?
25 benign \
42 malignant

```python
malignant_patient_ids = (
    df.groupby("PatientID")
      .filter(lambda g: (
          (g["AbnormalityType"] == "Mass") &
          (g["PatientGroundtruth"] == "Malignant")
      ).any())
      ["PatientID"]
      .unique()
)


benign_patient_ids = (
    df.groupby("PatientID")
      .filter(lambda g: (
          (g["AbnormalityType"] == "Mass") &
          (g["PatientGroundtruth"] == "Benign")
      ).any())
      ["PatientID"]
      .unique()
)




malignant_patient_ids
```

```python
df.filter(malignant_patient_ids)
```

### Breast density distribution

```python
mass_patients = df.loc[
    df["AbnormalityType"] == "Mass",
    "PatientID"
].unique()


patient_density = (
    df.loc[df["PatientID"].isin(mass_patients)]
      .groupby("PatientID")["BreastDensity"]
      .first()
)


density_counts = patient_density.value_counts()

density_counts
```
