```python
import os
import shutil
from pathlib import Path

# ===== CONFIG =====
root_dir = Path("datasets/CAMUS/")  # current structure root
output_root = Path("datasets/CAMUS_reformated")      # where new images/ and labels/ will be created
copy_files = True               # Set False if you want to MOVE instead of COPY

# ==================

new_images_dir = output_root / "images"
new_labels_dir = output_root / "labels"

new_images_dir.mkdir(exist_ok=True)
new_labels_dir.mkdir(exist_ok=True)

for patient_dir in root_dir.glob("patient*"):
    if not patient_dir.is_dir():
        continue

    patient_id = patient_dir.name  # e.g., patient001

    for file_path in patient_dir.iterdir():
        filename = file_path.name.lower()
        if (
        not file_path.is_file()
        or not filename.endswith(".nii.gz")
        or "half_sequence" in filename
        ):
            continue

        new_name = f"{file_path.name}"

        if "_gt" in file_path.name.lower():
            dest = new_labels_dir / new_name
        else:
            dest = new_images_dir / new_name

        if copy_files:
            shutil.copy2(file_path, dest)
        else:
            shutil.move(file_path, dest)

print("Reorganization complete.")
```
