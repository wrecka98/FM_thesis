```
from pathlib import Path
from typing import Callable, Optional, Sequence, Union

import numpy as np
import torch
from torch.utils.data import Dataset
import torch.nn.functional as F
import nibabel as nib


class MRIMiddleSlicesDataset(Dataset):
    """
    Load MRI .nii / .nii.gz volumes and return only the middle slice
    and its 2 adjacent slices as 2D images.

    For each volume, this dataset yields 3 samples:
        middle-1, middle, middle+1

    Supports:
    - 3D volumes: (H, W, D) or similar
    - 4D volumes: selects one time/frame first, then extracts slices

    Parameters
    ----------
    root : str or Path
        Root folder containing .nii or .nii.gz files.
    file_paths : optional sequence of paths
        Explicit list of files. If provided, `root` is ignored for scanning.
    slice_axis : int
        Axis along which to slice after reducing to 3D.
        Usually 2 for volumes shaped (H, W, D).
    time_index : str or int
        How to choose the frame if the input is 4D.
        - "middle": use middle time/frame
        - int: use that frame index
    resize_to : optional tuple[int, int]
        If given, resize each 2D slice to (H, W).
    repeat_channels : int
        Number of channels in output.
        - 1 => shape [1, H, W]
        - 3 => shape [3, H, W]
    transform : callable, optional
        Applied after tensor conversion.
    return_metadata : bool
        If True, returns a dict with file path, slice index, etc.
    """

    def __init__(
        self,
        root: Optional[Union[str, Path]] = None,
        file_paths: Optional[Sequence[Union[str, Path]]] = None,
        slice_axis: int = 2,
        time_index: Union[str, int] = "middle",
        resize_to: Optional[tuple[int, int]] = None,
        repeat_channels: int = 1,
        transform: Optional[Callable] = None,
        return_metadata: bool = False,
        label=0
    ):
        super().__init__()

        if file_paths is not None:
            self.files = [Path(p) for p in file_paths]
        else:
            if root is None:
                raise ValueError("Provide either `root` or `file_paths`.")
            root = Path(root)
            self.files = sorted(list(root.rglob("*.nii")) + list(root.rglob("*.nii.gz")))

        if len(self.files) == 0:
            raise ValueError("No .nii or .nii.gz files found.")

        self.slice_axis = slice_axis
        self.time_index = time_index
        self.resize_to = resize_to
        self.repeat_channels = repeat_channels
        self.transform = transform
        self.return_metadata = return_metadata
        self.label=label

        if repeat_channels not in (1, 3):
            raise ValueError("repeat_channels must be 1 or 3.")

        # Each volume contributes exactly 3 slices
        self.index_map = []
        for file_idx, path in enumerate(self.files):
            # We do not read the whole image here; just remember 3 logical slice slots.
            # slice_offset in {-1, 0, +1}
            for slice_offset in (-1, 0, 1):
                self.index_map.append((file_idx, slice_offset))

    def __len__(self) -> int:
        return len(self.index_map)

    def _load_nifti(self, path: Path) -> np.ndarray:
        vol = nib.load(str(path)).get_fdata()
        return np.asarray(vol, dtype=np.float32)

    def _reduce_to_3d(self, vol: np.ndarray) -> np.ndarray:
        """
        Reduce volume to 3D.
        - If 3D: return as is
        - If 4D: select one time/frame
        """
        if vol.ndim == 3:
            return vol

        if vol.ndim == 4:
            if self.time_index == "middle":
                t = vol.shape[-1] // 2
            elif isinstance(self.time_index, int):
                t = self.time_index
                if not (0 <= t < vol.shape[-1]):
                    raise IndexError(f"time_index={t} out of range for shape {vol.shape}")
            else:
                raise ValueError("time_index must be 'middle' or an integer")

            vol3d = vol[..., t]
            return np.asarray(vol3d, dtype=np.float32)

        raise ValueError(f"Expected 3D or 4D NIfTI, got shape {vol.shape}")

    def _normalize_mri_slice(self, img2d: np.ndarray) -> np.ndarray:
        """
        MRI-style normalization:
        - compute percentiles on nonzero voxels
        - clip to [0.5, 99.5] percentiles
        - scale to [0, 1]
        - keep background at 0
        """
        img = img2d.astype(np.float32)
        mask = img != 0

        if mask.sum() == 0:
            return np.zeros_like(img, dtype=np.float32)

        vals = img[mask]
        lo = np.percentile(vals, 0.5)
        hi = np.percentile(vals, 99.5)

        if hi <= lo:
            out = np.zeros_like(img, dtype=np.float32)
            out[mask] = 1.0
            return out

        img = np.clip(img, lo, hi)
        img = (img - lo) / (hi - lo)
        img[~mask] = 0.0
        return img.astype(np.float32)

    def _extract_slice(self, vol3d: np.ndarray, slice_offset: int) -> tuple[np.ndarray, int]:
        n_slices = vol3d.shape[self.slice_axis]
        mid = n_slices // 2
        idx = np.clip(mid + slice_offset, 0, n_slices - 1)

        img2d = np.take(vol3d, indices=idx, axis=self.slice_axis)
        return np.asarray(img2d, dtype=np.float32), int(idx)

    def _to_tensor(self, img2d: np.ndarray) -> torch.Tensor:
        x = torch.from_numpy(img2d).float().unsqueeze(0)  # [1, H, W]

        if self.resize_to is not None:
            x = F.interpolate(
                x.unsqueeze(0),
                size=self.resize_to,
                mode="bilinear",
                align_corners=False,
            ).squeeze(0)

        if self.repeat_channels == 3:
            x = x.repeat(3, 1, 1)

        return x
    
    def mask_to_bbox(mask):
        ys, xs = np.where(mask > 0)
        if len(xs) == 0:
            return None  # empty mask
        x_min, x_max = xs.min(), xs.max()
        y_min, y_max = ys.min(), ys.max()
        return [x_min, y_min, x_max, y_max]



    def __getitem__(self, idx: int):
        file_idx, slice_offset = self.index_map[idx]
        path = self.files[file_idx]

        vol = self._load_nifti(path)
        vol3d = self._reduce_to_3d(vol)

        img2d, slice_idx = self._extract_slice(vol3d, slice_offset)
        img2d = self._normalize_mri_slice(img2d)
        x = self._to_tensor(img2d)

        if self.transform is not None:
            x = self.transform(x)

        if self.return_metadata:
            meta = {
                "path": str(path),
                "file_name": path.name,
                "label":label,
                "slice_offset": slice_offset,
                "slice_index": slice_idx,
                "original_shape": tuple(vol.shape),
                "used_shape": tuple(vol3d.shape),
            }
            return x, self.label, meta

        return x,self.label
```

```
from pathlib import Path
from typing import Callable, Optional, Sequence, Union

import numpy as np
import torch
from torch.utils.data import Dataset
import torch.nn.functional as F
import nibabel as nib


class MRIMiddleSlicesROIDataset(Dataset):
    """
    Load MRI .nii / .nii.gz volumes and return only an ROI crop from:
      - middle-1 slice
      - middle slice
      - middle+1 slice

    ROI can be derived from:
      - a segmentation mask volume ("mask")
      - nonzero image foreground ("foreground")

    For each volume, this dataset yields 3 samples.

    Parameters
    ----------
    root : str or Path, optional
        Root folder containing image .nii / .nii.gz files.
    file_paths : optional sequence of paths
        Explicit list of image files. If provided, `root` is ignored.
    mask_root : str or Path, optional
        Root folder containing mask .nii / .nii.gz files.
    mask_file_paths : optional sequence of paths
        Explicit list of mask files aligned with `file_paths`.
    slice_axis : int
        Axis along which to slice after reducing to 3D.
    time_index : str or int
        If 4D, which frame to use. "middle" or integer.
    resize_to : optional tuple[int, int]
        Resize ROI crop to this size after cropping.
    repeat_channels : int
        1 or 3 output channels.
    transform : callable, optional
        Applied after tensor conversion.
    return_metadata : bool
        Whether to return metadata dict.
    label : int
        Label for all samples in this dataset.
    bbox_source : str
        "mask" or "foreground"
    bbox_margin : int
        Number of pixels to expand bbox on each side.
    bbox_square : bool
        If True, make bbox square before cropping.
    fallback_to_full_slice : bool
        If no bbox found, return full slice instead of error.
    """

    def __init__(
        self,
        root: Optional[Union[str, Path]] = None,
        file_paths: Optional[Sequence[Union[str, Path]]] = None,
        mask_root: Optional[Union[str, Path]] = None,
        mask_file_paths: Optional[Sequence[Union[str, Path]]] = None,
        slice_axis: int = 2,
        time_index: Union[str, int] = "middle",
        resize_to: Optional[tuple[int, int]] = None,
        repeat_channels: int = 1,
        transform: Optional[Callable] = None,
        return_metadata: bool = False,
        label: int = 0,
        bbox_source: str = "mask",          # "mask" or "foreground"
        bbox_margin: int = 10,
        bbox_square: bool = True,
        fallback_to_full_slice: bool = True,
    ):
        super().__init__()

        if file_paths is not None:
            all_files = [Path(p) for p in file_paths]
        else:
            if root is None:
                raise ValueError("Provide either `root` or `file_paths`.")
            root = Path(root)
            all_files = sorted(list(root.rglob("*.nii")) + list(root.rglob("*.nii.gz")))

        # keep only image files, exclude masks
        self.files = [
            p for p in all_files
            if not (p.name.endswith("_gt.nii") or p.name.endswith("_gt.nii.gz"))
        ]

        if len(self.files) == 0:
            raise ValueError("No image .nii or .nii.gz files found.")

        self.mask_files = None
        if bbox_source == "mask":
            self.mask_files = []
            for img_path in self.files:
                if img_path.name.endswith(".nii.gz"):
                    mask_path = img_path.with_name(img_path.name[:-7] + "_gt.nii.gz")
                elif img_path.name.endswith(".nii"):
                    mask_path = img_path.with_name(img_path.name[:-4] + "_gt.nii")
                else:
                    raise ValueError(f"Unsupported file extension: {img_path}")

                if not mask_path.exists():
                    raise FileNotFoundError(f"Mask file not found for image {img_path}: expected {mask_path}")

                self.mask_files.append(mask_path)

        self.slice_axis = slice_axis
        self.time_index = time_index
        self.resize_to = resize_to
        self.repeat_channels = repeat_channels
        self.transform = transform
        self.return_metadata = return_metadata
        self.label = label
        self.bbox_source = bbox_source
        self.bbox_margin = bbox_margin
        self.bbox_square = bbox_square
        self.fallback_to_full_slice = fallback_to_full_slice

        if repeat_channels not in (1, 3):
            raise ValueError("repeat_channels must be 1 or 3.")

        if bbox_source not in ("mask", "foreground"):
            raise ValueError("bbox_source must be 'mask' or 'foreground'.")

        # Each volume contributes exactly 3 slices
        self.index_map = []
        for file_idx in range(len(self.files)):
            for slice_offset in (-1, 0, 1):
                self.index_map.append((file_idx, slice_offset))

    def __len__(self) -> int:
        return len(self.index_map)

    def _load_nifti(self, path: Path) -> np.ndarray:
        vol = nib.load(str(path)).get_fdata()
        return np.asarray(vol, dtype=np.float32)

    def _reduce_to_3d(self, vol: np.ndarray) -> np.ndarray:
        if vol.ndim == 3:
            return vol

        if vol.ndim == 4:
            if self.time_index == "middle":
                t = vol.shape[-1] // 2
            elif isinstance(self.time_index, int):
                t = self.time_index
                if not (0 <= t < vol.shape[-1]):
                    raise IndexError(f"time_index={t} out of range for shape {vol.shape}")
            else:
                raise ValueError("time_index must be 'middle' or an integer")

            return np.asarray(vol[..., t], dtype=np.float32)

        raise ValueError(f"Expected 3D or 4D NIfTI, got shape {vol.shape}")

    def _normalize_mri_slice(self, img2d: np.ndarray) -> np.ndarray:
        img = img2d.astype(np.float32)
        mask = img != 0

        if mask.sum() == 0:
            return np.zeros_like(img, dtype=np.float32)

        vals = img[mask]
        lo = np.percentile(vals, 0.5)
        hi = np.percentile(vals, 99.5)

        if hi <= lo:
            out = np.zeros_like(img, dtype=np.float32)
            out[mask] = 1.0
            return out

        img = np.clip(img, lo, hi)
        img = (img - lo) / (hi - lo)
        img[~mask] = 0.0
        return img.astype(np.float32)

    def _extract_slice(self, vol3d: np.ndarray, slice_offset: int) -> tuple[np.ndarray, int]:
        n_slices = vol3d.shape[self.slice_axis]
        mid = n_slices // 2
        idx = np.clip(mid + slice_offset, 0, n_slices - 1)

        img2d = np.take(vol3d, indices=idx, axis=self.slice_axis)
        return np.asarray(img2d, dtype=np.float32), int(idx)

    @staticmethod
    def mask_to_bbox(mask: np.ndarray):
        ys, xs = np.where(mask > 0)
        if len(xs) == 0:
            return None
        x_min, x_max = xs.min(), xs.max()
        y_min, y_max = ys.min(), ys.max()
        return [int(x_min), int(y_min), int(x_max), int(y_max)]
    
    @staticmethod
    def _is_mask_file(path: Path) -> bool:
        return path.name.endswith("_gt.nii") or path.name.endswith("_gt.nii.gz")

    @staticmethod
    def _image_to_mask_path(path: Path) -> Path:
        name = path.name
        if name.endswith(".nii.gz"):
            return path.with_name(name[:-7] + "_gt.nii.gz")
        elif name.endswith(".nii"):
            return path.with_name(name[:-4] + "_gt.nii")
        else:
            raise ValueError(f"Unsupported file extension: {path}")

    def _expand_bbox(self, bbox, h: int, w: int):
        x_min, y_min, x_max, y_max = bbox

        x_min -= self.bbox_margin
        y_min -= self.bbox_margin
        x_max += self.bbox_margin
        y_max += self.bbox_margin

        x_min = max(0, x_min)
        y_min = max(0, y_min)
        x_max = min(w - 1, x_max)
        y_max = min(h - 1, y_max)

        if self.bbox_square:
            bw = x_max - x_min + 1
            bh = y_max - y_min + 1
            side = max(bw, bh)

            cx = (x_min + x_max) / 2.0
            cy = (y_min + y_max) / 2.0

            x_min = int(round(cx - side / 2))
            x_max = x_min + side - 1
            y_min = int(round(cy - side / 2))
            y_max = y_min + side - 1

            # shift back into image if out of bounds
            if x_min < 0:
                x_max -= x_min
                x_min = 0
            if y_min < 0:
                y_max -= y_min
                y_min = 0
            if x_max >= w:
                shift = x_max - (w - 1)
                x_min -= shift
                x_max = w - 1
            if y_max >= h:
                shift = y_max - (h - 1)
                y_min -= shift
                y_max = h - 1

            x_min = max(0, x_min)
            y_min = max(0, y_min)
            x_max = min(w - 1, x_max)
            y_max = min(h - 1, y_max)

        return [x_min, y_min, x_max, y_max]

    def _crop_from_bbox(self, img2d: np.ndarray, bbox):
        x_min, y_min, x_max, y_max = bbox
        return img2d[y_min:y_max + 1, x_min:x_max + 1]

    def _to_tensor(self, img2d: np.ndarray) -> torch.Tensor:
        x = torch.from_numpy(img2d).float().unsqueeze(0)  # [1, H, W]

        if self.resize_to is not None:
            x = F.interpolate(
                x.unsqueeze(0),
                size=self.resize_to,
                mode="bilinear",
                align_corners=False,
            ).squeeze(0)

        if self.repeat_channels == 3:
            x = x.repeat(3, 1, 1)

        return x

    def __getitem__(self, idx: int):
        file_idx, slice_offset = self.index_map[idx]
        path = self.files[file_idx]

        vol = self._load_nifti(path)
        vol3d = self._reduce_to_3d(vol)
        img2d, slice_idx = self._extract_slice(vol3d, slice_offset)

        bbox = None

        if self.bbox_source == "mask":
            mask_path = self.mask_files[file_idx]
            mask_vol = self._load_nifti(mask_path)
            mask_vol3d = self._reduce_to_3d(mask_vol)
            mask2d, _ = self._extract_slice(mask_vol3d, slice_offset)

            # Any positive label is treated as ROI
            mask_bin = (mask2d > 0).astype(np.uint8)
            bbox = self.mask_to_bbox(mask_bin)

        elif self.bbox_source == "foreground":
            fg_mask = (img2d != 0).astype(np.uint8)
            bbox = self.mask_to_bbox(fg_mask)

        h, w = img2d.shape

        if bbox is None:
            if self.fallback_to_full_slice:
                roi2d = img2d
                used_bbox = [0, 0, w - 1, h - 1]
            else:
                raise ValueError(f"No ROI found for file {path}, slice {slice_idx}")
        else:
            used_bbox = self._expand_bbox(bbox, h=h, w=w)
            roi2d = self._crop_from_bbox(img2d, used_bbox)

        roi2d = self._normalize_mri_slice(roi2d)
        x = self._to_tensor(roi2d)

        if self.transform is not None:
            x = self.transform(x)

        if self.return_metadata:
            meta = {
                "path": str(path),
                "file_name": path.name,
                "label": self.label,
                "slice_offset": slice_offset,
                "slice_index": slice_idx,
                "original_shape": tuple(vol.shape),
                "used_shape": tuple(vol3d.shape),
                "bbox_xyxy": used_bbox,
                "roi_shape": tuple(roi2d.shape),
                "bbox_source": self.bbox_source,
            }
            return x, self.label, meta

        return x, self.label
```

```
# DINOv2-base is the closest Hugging Face DINOv2 size to a ViT-B / medSAM-vit_b backbone.
DINO_MODEL_NAME = "facebook/dinov2-base"
DINO_IMAGE_SIZE = 224
DINO_PATCH_SIZE = 14
DINO_EMBED_DIM = 768

dinov2 = AutoModel.from_pretrained(DINO_MODEL_NAME).to(device).eval()
print("Loaded:", DINO_MODEL_NAME)

DINO_MEAN = torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1)
DINO_STD = torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1)

def preprocess_for_dinov2(x: torch.Tensor) -> torch.Tensor:
    """
    x: [B, C, H, W]
    - accepts float tensors in [0, 1] or image-like tensors in [0, 255]
    - converts grayscale to 3 channels
    - resizes to 224x224
    - normalizes with ImageNet mean/std as expected by DINOv2
    """
    if not torch.is_floating_point(x):
        x = x.float()

    if x.max() > 1.0:
        x = x / 255.0

    if x.shape[1] == 1:
        x = x.repeat(1, 3, 1, 1)
    elif x.shape[1] > 3:
        x = x[:, :3, :, :]

    x = F.interpolate(
        x,
        size=(DINO_IMAGE_SIZE, DINO_IMAGE_SIZE),
        mode="bicubic",
        align_corners=False,
    )

    mean = DINO_MEAN.to(device=x.device, dtype=x.dtype)
    std = DINO_STD.to(device=x.device, dtype=x.dtype)
    x = (x - mean) / std
    return x

@torch.no_grad()
def extract_dinov2_features(x: torch.Tensor):
    """
    Returns:
      cls_features:   [B, 768]
      patch_features: [B, 768, 16, 16]
      patch_tokens:   [B, 256, 768]
    """
    x = preprocess_for_dinov2(x).to(device)
    outputs = dinov2(pixel_values=x)
    hidden = outputs.last_hidden_state
    cls_features = hidden[:, 0]
    patch_tokens = hidden[:, 1:]

    grid = DINO_IMAGE_SIZE // DINO_PATCH_SIZE
    patch_features = patch_tokens.transpose(1, 2).reshape(-1, DINO_EMBED_DIM, grid, grid)
    return cls_features, patch_features, patch_tokens

@torch.no_grad()
def get_embedding(batch_images: torch.Tensor) -> torch.Tensor:
    cls_features, _, _ = extract_dinov2_features(batch_images)
    return cls_features
```

```
datasets_path = "../../datasets/"
prep_datasets = "/prep_datasets"
ID_path = datasets_path + prep_datasets + "/MnM_splits/vendor_split/Philips"
OOD_path = datasets_path + prep_datasets + "/MnM_splits/vendor_split/GE"

# Use 224x224 because DINOv2-base is trained around this input size.
id_roi_dataset = MRIMiddleSlicesROIDataset(
    root=ID_path,
    mask_root=ID_path,
    bbox_source="mask",
    resize_to=(224, 224),
    repeat_channels=3,
    label=0,
    return_metadata=True,
)

ood_roi_dataset = MRIMiddleSlicesROIDataset(
    root=OOD_path,
    mask_root=OOD_path,
    bbox_source="mask",
    resize_to=(224, 224),
    repeat_channels=3,
    label=-1,
    return_metadata=True,
)

id_roi_loader = DataLoader(id_roi_dataset, batch_size=4, shuffle=False, num_workers=0)
ood_roi_loader = DataLoader(ood_roi_dataset, batch_size=4, shuffle=False, num_workers=0)

print("ID samples:", len(id_roi_dataset))
print("OOD samples:", len(ood_roi_dataset))
```

```
# Stage 1: extract DINOv2 embeddings from the ID data and fit Mahalanobis

feats_list = []
labels_list = []

dinov2.eval()
with torch.no_grad():
    for imgs, y, _ in id_roi_loader:
        imgs = imgs.to(device)
        y = y.to(device)

        z = get_embedding(imgs)   # [B, 768] CLS token
        feats_list.append(z)
        labels_list.append(y)

Z_id = torch.cat(feats_list, dim=0)
Y_id = torch.cat(labels_list, dim=0)

print("Z_id:", Z_id.shape)
print("Y_id:", Y_id.shape, "unique labels:", torch.unique(Y_id))

detector = Mahalanobis(model=None)
detector.fit_features(Z_id, Y_id)
```

```
# Predict on OOD dataset

feats_list = []
with torch.no_grad():
    for imgs, _, _ in ood_roi_loader:
        imgs = imgs.to(device)
        z = get_embedding(imgs)
        feats_list.append(z)

Z_ood = torch.cat(feats_list, dim=0)

scores_id = detector.predict_features(Z_id)
scores_ood = detector.predict_features(Z_ood)

print("Z_ood:", Z_ood.shape)
print("scores_id:", scores_id.shape)
print("scores_ood:", scores_ood.shape)
```

```
# Evaluate OOD metrics

y_id = torch.zeros(len(scores_id), dtype=torch.long)
y_ood = -torch.ones(len(scores_ood), dtype=torch.long)

scores = torch.cat([scores_id.detach().cpu(), scores_ood.detach().cpu()], dim=0)
y_true = torch.cat([y_id, y_ood], dim=0)

metrics = OODMetrics()
metrics.update(scores, y_true)
result = metrics.compute()

print(result)
```
